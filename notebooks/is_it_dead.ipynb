```python
import duckdb
```

# Current database

```python
con = duckdb.connect()
for ext in ["postgres", "spatial"]:
    con.install_extension(ext)
    con.load_extension(ext)
con.sql("""
    ATTACH 'dbname=spartid_ais user=postgres password=postgres host=127.0.0.1' AS db (TYPE postgres);
""")
```

```python
con.sql("SELECT * FROM db.public.last_position ORDER BY timestamp DESC").df()
```

```python
con.sql("SELECT * FROM db.historic_position WHERE timestamp < '2025-04-08' AND timestamp > '2025-04-07';")
```
